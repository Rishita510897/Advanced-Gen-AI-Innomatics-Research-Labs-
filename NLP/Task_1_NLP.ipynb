```python
# Task-2
import re
def preprocess_text(text):
  try:
    if not text or not isinstance(text,str):
      return [],""
    text=text.lower() # convert to lowercase
    text=re.sub(r'http\S+|www\S+','',text) # remove urls
    text=re.sub(r'\S+@\S+','',text) # remove emails
    text=re.sub(r'\d+','',text) # remove numbers
    text=re.sub(r'(.)\1{2,}',r'\1',text) # handle repeated characters
    text=re.sub(r'[^a-z\s]','',text) # remove special characters
    text=re.sub(r'\s+','',text).strip() # remove extra spaces
    tokens=text.split() #tokenize
    filtered_tokens=[
        word for word in tokens
        if len(word)>2 or word in ["no","not"]
    ]
    clean_sentence=" ".join(filtered_tokens)
    return filtered_tokens,clean_sentence
  except Exception as e:
    return [],""
```

```python
# Task-3
test_sentences = [
    "Get 100% FREE access now!!!",
    "I absolutely looooved this product 😍😍",
    "Worst service ever... 0/10",
    "Call me at 9876543210",
    "This is THE best course!!!",
    "Visit https://openai.com now!",
    "Nooooo this is baaad!!!",
    "OK OK OK I got it",
    "Win $$$ now!!! Limited offer!!!",
    "I am not happy with this"
]
results=[]
for text in test_sentences:
  tokens,clean_sentence=preprocess_text(text)
  print("\nOriginal:",text)
  print("Tokens:",tokens)
  print("Cleaned Sentence:",clean_sentence)
  results.append((tokens,clean_sentence))
```

```python
# Task-4
import numpy as np
analytics=[]
for i,(tokens,_) in enumerate(results):
  if tokens:
    total_tokens=len(tokens)
    unique_tokens=len(set(tokens))
    avg_length=np.mean([len(word) for word in tokens])
  else:
    total_tokens=unique_tokens=avg_length=0
  analytics.append((i,total_tokens,unique_tokens,avg_length))
  print(f"\nSentence {i+1}:")
  print("Total Tokens:",total_tokens)
  print("Unique Tokens:",unique_tokens)
  print("Average Length:",round(avg_length,2))
```

=> Most noisy sentence: "Get 100% FREE access now!!!(contains numbers, symbols, uppercase, punctuation)
=> Most meaningful after cleaning: "I am not happy with this"(retains sentiment and important words like not, happy)

```python
# Task-5
from collections import Counter
all_tokens=[]
for tokens,_ in results:
  all_tokens.extend(tokens)
counter=Counter(all_tokens)
print("\nTop 10 frequent words:")
print(counter.most_common(10))
print("\nTop 5 least frequent words:")
print(counter.most_common()[:-6:-1])
```

```python
# Task-6
def full_pipeline(text_list):
  all_tokens=[]
  clean_sentences=[]
  for text in text_list:
    tokens,clean_sentence=preprocess_text(text)
    all_tokens.extend(tokens)
    clean_sentences.append(clean_sentence)
  return {
      "tokens":all_tokens,
      "clean_sentences":clean_sentences
  }
output=full_pipeline(test_sentences)
print(output)
```
